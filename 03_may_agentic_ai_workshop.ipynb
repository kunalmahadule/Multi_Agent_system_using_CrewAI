```python
# import libs


import os
import gradio as gr
from crewai import Agent, Task, Crew
from crewai_tools import SerperDevTool


os.environ["SERPER_API_KEY"] = "2138bc983d4c1605b3c64718675f8cab788adaa9"
os.environ["GOOGLE_API_KEY"] = "AIzaSyBh7uOeHyvMyekanwIBCoOwczEwlrWVv_E"

search_tool = SerperDevTool()

def run_multi_agent(topic):

  # Creating different agents

  teacher = Agent(role="Teacher",
                  goal = 'Explain concepts clearly step by step',
                  backstory = 'Experienced teacher who explain with example',
                  llm='gemini/gemini-2.5-flash',
                  verbose = False
                  )


  researcher = Agent(role = 'Researcher',
                    goal = 'Find latest information',
                    backstory= 'Expert in searching real-world data',
                    tools = [search_tool],
                    llm='gemini/gemini-2.5-flash',
                    verbose = False
                     )


  simplifier = Agent(role = 'Simplifier',
                    goal = 'Make thing easy',
                    backstory = 'Breaks complex ideas into simple language',
                    llm='gemini/gemini-2.5-flash',
                    verbose = False
                     )


  student = Agent(role = 'Student',
                  goal = "Take notes",
                  backstory = 'write a simple notes like learner',
                  llm='gemini/gemini-2.5-flash',
                  verbose = False
                  )

  examiner = Agent(role = 'Examiner',
                  goal= 'Create questions',
                  backstory = 'Tests understanding',
                  llm='gemini/gemini-2.5-flash',
                  verbose = False
                   )


  # topic = 'What is difference between agents vs agentic ai vs mcp vs generative ai'

  # Assigning different tasks to different agents

  task1 = Task(
  description = f'Explain {topic} in simple steps with examples',
  expected_output = 'I need step by step explaination',
  agent = teacher
  )

  task2 = Task(
  description = f'search and find 3 important points about {topic}',
  expected_output = '3 clear points',
  agent = researcher
  )


  task3 = Task(
  description = f'Simplify the explaination of the {topic}',
  expected_output = 'very simple explainations',
  agent = simplifier
  )


  task4 = Task(
  description = f'Write short notes on {topic} like a student',
  expected_output = 'Simple notes',
  agent = student
  )


  task5 = Task(
  description = f'create 5 question about {topic}',
  expected_output = 'A list of 5 clear and challenging multiple-choice questions with their correct answers.',
  agent = examiner
  )


  # MultiAgent varible initialization

  crew = Crew(
      agents=[teacher, researcher, simplifier, student, examiner],
      tasks=[task1, task2, task3, task4, task5],
      verbose=False
  )


  result = crew.kickoff()

  return str(result)


interface = gr.Interface(
    fn = run_multi_agent,
    inputs = gr.Textbox(
    label = "Enter Topic",
    placeholder="what is Agentic ai vs single "
    ),

    outputs = gr.Textbox(
        label= "Multi-Agent Output", lines=30),

    title = "Muli-Agent System using CrewAI",
    description= "Enter any topic and see multiple AI agents colaborate to teach, simplify and test you."

)

if __name__ == "__main__":
  interface.launch(share=True)
```
